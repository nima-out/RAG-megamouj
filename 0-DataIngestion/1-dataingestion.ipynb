Text files (simple .txt)

```python
import os

os.makedirs('data/text_files',exist_ok= True)
```

```python
sample_texts = {
    "data/text_files/python_intro.txt": """   
    In the world of software development, writing clean and 
    readable code is a skill that distinguishes a novice from a 
    professional. Utilizing object-oriented principles, adhering to 
    PEP 8 standards in Python, and writing unit tests for Django projects
    not only simplifies future maintenance but also enables other team members
    to grasp the program's logic more quickly. Always remember that your code 
    is written more for humans to read than for computers to execute.  
    """
}


for filepath,content in sample_texts.items():
    with open(filepath,'w',encoding="utf-8") as f:
        f.write(content)
        
print("sample file created!")
```

## text loaders read single files -> TEXT LOADERS

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("data/text_files/python_intro.txt")

documents = loader.load()
```

## Directory loader read multiple text files -> DIRECTORY Loader

```python
sample_texts = {
    "data/text_files/daw.txt": """   
    The evolution of music technology has transformed how we approach creativity. 
    With advancements in digital signal processing, guitarists can now replicate 
    the nuanced response of high-end tube amplifiers using software like THU, 
    making it possible to achieve studio-quality tones at home. Whether you are 
    experimenting with extended-range instruments like 7 or 8-string guitars or 
    layering complex arrangements in a DAW, the barrier between a raw idea and a
    polished final track continues to vanish, placing the power of a professional 
    production house directly into the hands of the individual artist.
    """
}


for filepath,content in sample_texts.items():
    with open(filepath,'w',encoding="utf-8") as f:
        f.write(content)
        
print("sample file created!")
```

```python
from langchain_community.document_loaders import DirectoryLoader


dir_loader = DirectoryLoader(
    "data/text_files",
    glob ="**/*.txt",
    loader_cls= TextLoader,
    loader_kwargs={'encoding': 'utf-8'},
    show_progress= True
)

documents = dir_loader.load()
```

## TEXT splititng strategies

```python
from langchain_text_splitters import(
    RecursiveCharacterTextSplitter,
    CharacterTextSplitter,
    TokenTextSplitter
)

print(documents)
```

### METHOD 1 - Character text splitter

```python
text = documents[0].page_content

char_splitter = CharacterTextSplitter(
    
    separator="\n",
    chunk_size = 150,
    chunk_overlap = 20,
    length_function =   len
)


char_chunks = char_splitter.split_text(text)
print(f"created{len(char_chunks)} chunks")
print(f"first chunk : {char_chunks[0][:100]}")
```

```python
print(char_chunks[1])
print('----------')
print(char_chunks[2])
```

### METHOD 2 - Recurssive text splitter

```python
text = documents[0].page_content

recursive_splitter  = RecursiveCharacterTextSplitter(
    
    separators=["\n",],
    chunk_size = 150,
    chunk_overlap = 20,
    length_function =   len
)


recursive_chunks = recursive_splitter.split_text(text)
```

```python
print(recursive_chunks[0])
print("------------")
print(recursive_chunks[1]) 

print(recursive_chunks[3])
```
