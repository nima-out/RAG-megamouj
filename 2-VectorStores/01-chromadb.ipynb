```python
## save sample doc to files

import tempfile
temp_dir = tempfile.mkdtemp()

for i, doc in enumerate(sample_docs):
    with open(f"{temp_dir}/doc{i}.txt",'w',encoding="utf-8") as f:
        f.write(doc)
        

print(f"file created at : {temp_dir}")
```

```python
## document loading
from langchain_community.document_loaders import DirectoryLoader,TextLoader

loader = DirectoryLoader(
    temp_dir,
    glob="*.txt",
    loader_cls= TextLoader,
    loader_kwargs={"encoding":"utf_8"}
)

document = loader.load()

print(f"loaded {len(document)} document")
print("\n first document preview:")
print(document[0].page_content[:200] + "...")
```

```python
### document splitter


text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 500,
    chunk_overlap = 50,
    length_function = len,
    #separators= ["\n\n","\n",". "," ",""]
    separators= [" "]
)

chunks = text_splitter.split_documents (document)



```

```python
chunks
```

```python
### embedding models

embeddings = HuggingFaceEmbeddings(model_name = "sentence-transformers/all-MiniLM-L6-v2")
embeddings
```

```python
#testing the vector embedding

sample_text = "sun is shinning!"
vector = embeddings.embed_query(sample_text)
vector
```

## initialaze the chroma DB vector store and store the cunks there in vector representation

```python
## create a chroma db vector store

persist_directory ="./chroma_db"

## initialize Chromadb with huggingface embeddings
vectorstore = Chroma.from_documents(
    documents = chunks,
    embedding = HuggingFaceEmbeddings(),
    persist_directory=persist_directory,
    collection_name= "rag_collection"
    
)


print(vectorstore._collection.count())
print("persist_directory")
```

```python
## test similarity search

query = "what are the types of machine learnign"

similar_docs = vectorstore.similarity_search(query,k=3)
similar_docs
```

```python
query = "what is nlp"

similar_docs = vectorstore.similarity_search(query,k=3)
similar_docs
```

```python
query = "what is Deep learning"
similar_docs = vectorstore.similarity_search(query,k=3)
similar_docs
```
